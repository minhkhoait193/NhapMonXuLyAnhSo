```python
import numpy as np
import imageio.v2 as iio
import matplotlib.pyplot as plt
from scipy import ndimage as nd

# Load the image
data = iio.imread('exercise/mamdoan.jpg')

canh_chua = data[82:369, 283:578]  # Example values; change them after testing



# Shift the kiwi 30 pixels to the right (down, right)
canh_chua = nd.shift(canh_chua, shift=(0, 30, 0))  # (row, column, channel)

# Display the result
plt.imshow(canh_chua)
plt.axis('off')
plt.title("canh chua shifted 30 pixels to the right")
plt.show()



```

```python
import numpy as np
import imageio.v2 as iio
import matplotlib.pyplot as plt
from scipy import ndimage as nd

# Đọc ảnh
data = iio.imread('exercise/mamdoan.jpg')

# Phóng to ảnh gấp 3 lần
zoom_in = nd.zoom(data, (3, 3, 1))  

# Thu nhỏ ảnh 0.3 lần
zoom_out = nd.zoom(data, (0.3, 0.3, 1))

# Hiển thị kết quả
plt.figure(figsize=(10, 4))

plt.subplot(1, 3, 1)
plt.imshow(data)
plt.title("Ảnh gốc")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(zoom_in)
plt.title("Phóng to 3 lần")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(zoom_out)
plt.title("Thu nhỏ 0.3 lần")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import imageio.v2 as iio
import matplotlib.pyplot as plt
from scipy import ndimage as nd

# Đọc ảnh
data = iio.imread('exercise/mamdoan.jpg')

# Xoay 45 độ với reshape=True (mặc định)
rotated_true = nd.rotate(data, 45, reshape=True)

# Xoay 45 độ với reshape=False (giữ nguyên kích thước ban đầu)
rotated_false = nd.rotate(data, 45, reshape=False)

# Hiển thị kết quả
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(data)
plt.title("Ảnh gốc")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(rotated_true)
plt.title("Xoay 45° (reshape=True)")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(rotated_false)
plt.title("Xoay 45° (reshape=False)")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import scipy.ndimage as nd
import imageio.v2 as iio
import matplotlib.pyplot as plt

def GeoFun(outcoord):
    a = 10 * np.cos(outcoord[0]/10.0) + outcoord[0]
    b = 10 * np.cos(outcoord[1]/10.0) + outcoord[1]
    return a, b

# Đọc ảnh
data = iio.imread('exercise/mamdoan.jpg', mode = 'F')

# Áp dụng biến đổi
d1 = nd.geometric_transform(data, GeoFun)
plt.imshow(d1, cmap='gray')
plt.show()
```

```python
import numpy as np
import imageio.v2 as iio
import matplotlib.pyplot as plt
from scipy import ndimage as nd

# Đọc ảnh grayscale
data = iio.imread('exercise/mamdoan.jpg', mode='F')

# Áp dụng phép Dilation và Erosion với size=(5, 5)
structure = np.ones((5, 5))  # kernel 5x5

dilated = nd.binary_dilation(data, structure=structure)
eroded = nd.binary_erosion(data, structure=structure)

# Hiển thị kết quả
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(data, cmap='gray')
plt.title("Ảnh gốc")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(dilated, cmap='gray')
plt.title("Dilation (5x5)")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(eroded, cmap='gray')
plt.title("Erosion (5x5)")
plt.axis('off')

plt.tight_layout()
plt.show()
```
